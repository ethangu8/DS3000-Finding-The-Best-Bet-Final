```python
away = list(h2h_combined['away_team'].unique())
home = list(h2h_combined['home_team'].unique())

teams = list(set(away+home))
```

```python
teams
```

```python
wins_dict = dict()

wins = [46, 44, 42, 44, 52, 51, 48, 53, 46, 53, 43, 51, 64, 49, 48, 51, 36, 56]
home_win_p = [68.2, 48.8, 59.5, 49.5, 65.1, 55.8, 63.6, 76.7, 55.8, 72.1, 65.9, 60.5, 76.7, 69.8, 58.1, 69.1, 47.7, 72.1]
away_win_p = [46.5, 55.8, 40.5, 45.2, 60.5, 58.5, 48.8, 53.5, 58.1, 58.1, 38.6, 65.1, 76.7, 48.8, 55.8, 57.1, 43.2, 62.8] 
combined = dict(zip(teams, wins,))
win_p = dict(zip(home_win_p, away_win_p))
```

```python
h2h_combined['away_wins'] = h2h_combined['away_team'].map(combined)
h2h_combined['home_wins'] = h2h_combined['home_team'].map(combined)
h2h_combined['home_win_p'] = h2h_combined['home_team'].map(win_p)
h2h_combined['away_win_p'] = h2h_combined['away_team'].map(win_p)
```

```python
h2h_combined
```

```python
from sklearn.cluster import KMeans
import seaborn as sns
import matplotlib.pyplot as plt

# k in k-means describes the number of cliusters
n_clusters = 2

#extract x features (define distance between samples)
x_feat_list = ['home_wins']
x = h2h_combined.loc[:, x_feat_list].values

#perform clustering 
kmeans = KMeans(n_clusters=n_clusters)
kmeans.fit(x)
y = kmeans.predict(x)

sns.scatterplot(data=h2h_combined, x='home_team', y='home_wins', s=100, hue=y, palette='Set2')
plt.gcf().set_size_inches(12, 6)
plt.suptitle('Home wins')
```
